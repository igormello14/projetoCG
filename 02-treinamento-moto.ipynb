```python
from ultralytics import YOLO

# Carrega o modelo base "nano" que usamos antes
model = YOLO('yolov8n.pt')

# Inicia o treinamento com os dados customizados
resultados = model.train(
    data='./Moto.v1i.yolov8/data.yaml', # Ajuste o caminho se a sua pasta tiver outro nome
    epochs=5,                       # Número de vezes que o modelo vai ver todas as imagens
    imgsz=640,                       # Resolução padrão do YOLO
    batch=8,                         # Quantas imagens processar por vez (ajuste se faltar memória)
    name='detector_motociclista'     # Nome da pasta onde seu novo modelo será salvo
)
```

```python
import glob
import os

# Procura todas as pastas de resultado criadas pelo YOLO
pastas_runs = glob.glob('runs/detect/*')

if pastas_runs:
    # Filtra para encontrar a pasta gerada mais recentemente
    ultima_pasta = max(pastas_runs, key=os.path.getctime)
    print(f"O seu último treinamento foi salvo na pasta: {ultima_pasta}")
    
    caminho_modelo = f"{ultima_pasta}/weights/best.pt"
    print(f"\\nCopie este caminho exato para usar na inferência:")
    print(f"'{caminho_modelo}'")
else:
    print("Nenhuma pasta de treinamento foi encontrada. Verifique se você está no diretório correto do projeto_final.")
```
